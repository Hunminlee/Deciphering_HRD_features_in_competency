# Generalization Test

- Train on dataset 2020-2021 and 2021-2021 combined
- Test on dataset 2022-2023 and 2023-2023

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

def remove_prefix(df, year_prefix):
    """Removes the Wxx prefix from column names."""
    prefix = f"W{year_prefix[-2:]}"  # e.g., 'W20' from '2020'
    new_columns = {col: col[len(prefix):] for col in df.columns if col.startswith(prefix)}
    return df.rename(columns=new_columns)

# Step 1: Remove prefixes
df_20_21_clean = remove_prefix(df_20_21, '2020')
df_21_21_clean = remove_prefix(df_21_21, '2021')
df_21_23_clean = remove_prefix(df_21_23, '2021')
df_22_23_clean = remove_prefix(df_22_23, '2022')
df_23_23_clean = remove_prefix(df_23_23, '2023')

# Step 2: Get common columns
common_cols = list(
    set(df_20_21_clean.columns)
    & set(df_21_21_clean.columns)
    & set(df_22_23_clean.columns)
    & set(df_23_23_clean.columns)
)

# Step 3: Align columns / Ensure y values are Series
X_train = pd.concat([df_20_21_clean[common_cols], df_21_21_clean[common_cols]], axis=0).reset_index(drop=True)
y_train = pd.concat([pd.Series(y_20_21), pd.Series(y_21_21)], axis=0).reset_index(drop=True)

X_test_1 = df_22_23_clean[common_cols]
y_test_1 = pd.Series(y_22_23)

X_test_2 = df_23_23_clean[common_cols]
y_test_2 = pd.Series(y_23_23)


# Step 5: Train model
model_instance = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
model_instance.fit(X_train, y_train)

# Step 6: Predict
y_pred = model_instance.predict(X_test_1)
acc_generalization = accuracy_score(y_test_1, y_pred)
print(f"📊 Generalization Accuracy (2020/2021 → 2022): {acc_generalization:.4f}")

y_pred = model_instance.predict(X_test_2)
acc_generalization = accuracy_score(y_test_2, y_pred)
print(f"📊 Generalization Accuracy (2020/2021 → 2023): {acc_generalization:.4f}")
```
